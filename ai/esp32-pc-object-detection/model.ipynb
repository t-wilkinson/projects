```python
import albumentations as A
import cv2
import matplotlib.pyplot as plt
from pycocotools.coco import COCO
```

```python
transform = A.Compose([
    A.Resize(240, 240),
    A.RandomBrightnessContrast(p=0.5),
    A.GaussNoise(var_limit=(100, 500), p=0.4), # Crucial for ESP32 sensors
    A.Blur(blur_limit=3, p=0.3),               # Simulates out-of-focus or motion
    A.HueSaturationValue(p=0.3),
], bbox_params=A.BboxParams(format='coco', label_fields=['category_ids']))
```

```python
def preprocess_example(image_path, bboxes, category_ids):
    image = cv2.imread(image_path)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    
    transformed = transform(image=image, bboxes=bboxes, category_ids=category_ids)
    
    plt.imshow(transformed['image'])
    plt.title("Simulated ESP32-CAM View")
    plt.show()
    return transformed['image'], transformed['bboxes']
```
